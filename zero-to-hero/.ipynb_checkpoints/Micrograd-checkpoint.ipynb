```python
from collections import deque
import numpy as np
import random



class Value():
    def __init__(self, data, children=[], op=None, label=''):
        self.data = data
        self.children = children
        self.op = op
        self.label = label
        self.grad = 0.0
        # self._backward = None

    def __add__(self, other):
        other = other if isinstance(other, Value) else Value(other)
        out = Value(self.data + other.data, [self, other], '+')

        def _backward():
            self.grad += out.grad
            other.grad += out.grad
        
        out._backward = _backward
        return out

    def __pow__(self, other):
        assert isinstance(other, (int, float)), "only supporting int/float powers for now"
        out = Value(self.data**other, (self,), f'**{other}')

        def _backward():
            self.grad += (other * self.data**(other-1)) * out.grad
        out._backward = _backward

        return out

    def __rmul__(self, other): # other * self
        return self * other

    def __truediv__(self, other): # self / other
        return self * other**-1

    def __rtruediv__(self, other): # other / self
        return other * self**-1
    
    def __neg__(self): # -self
        return self * -1

    def __radd__(self, other): # other + self
        return self + other

    def __sub__(self, other): # self - other
        return self + (-other)

    def __rsub__(self, other): # other - self
        return other + (-self)
        
    def __mul__(self, other):
        other = other if isinstance(other, Value) else Value(other)
        out = Value(self.data * other.data, [self, other], '*')

        def _backward():
            self.grad += other.data * out.grad
            other.grad += self.data * out.grad

        out._backward = _backward
        return out

    def sigmoid(self):
        out = Value(1 / (1 + np.exp(-self.data)), [self], 'sigmoid')

        def _backward():
            self.grad += out.data * (1 - out.data) * out.grad

        out._backward = _backward
        return out

    def swish(self):
        sig = 1 / (1 + np.exp(-self.data))
        out = Value(sig * self.data, [self], 'swish')

        def _backward():
            self.grad += (sig + sig * (1 - sig) * self.data) * out.grad
            
        out._backward = _backward
        return out
        
        
    def __repr__(self):
        return f"Value(data={self.data},children={len(self.children)},op={self.op})"

    def _backward(self):
        pass # leaf node
        # raise Exception("Not implemented")       

    def backward(self):
        # count how many edges lead into each node from above (its consumers),
        # i.e. how many gradient contributions it has to wait for
        pending = {}
        seen = {self}
        queue = deque([self])
        while queue:
            node = queue.popleft()
            for child in node.children:
                pending[child] = pending.get(child, 0) + 1
                if child not in seen:
                    seen.add(child)
                    queue.append(child)

        # zero every grad below the root so re-running backward doesn't accumulate
        for node in seen:
            if node is not self:
                node.grad = 0.0

        # BFS from the root, but only enqueue a node once all its consumers have
        # pushed their gradient into it (Kahn's algorithm), so its grad is final
        queue = deque([self])
        while queue:
            node = queue.popleft()
            node._backward()
            for child in node.children:
                pending[child] -= 1
                if pending[child] == 0:
                    queue.append(child)



a = Value(3.0); a.label='a'
b = Value(2.0); b.label='b'
bs = b.swish(); bs.label = 'bs'
c = a * bs; c.label='c'
f = Value(3.0); f.label='f'
e = c * f; e.label='e'
g  = e + f; g.label='g'
h = g.sigmoid(); h.label = 'h'

g
```

```python
from graphviz import Digraph

def trace(root):
    # builds a set of all nodes and edges in a graph
    nodes, edges = set(), set()
    def build(v):
        if v not in nodes:
            nodes.add(v)
            for child in v.children:
                edges.add((child, v))
                build(child)
    build(root)
    return nodes, edges

def draw_dot(root):
    dot = Digraph(format='svg', graph_attr={'rankdir': 'LR'}) # LR = left to right

    nodes, edges = trace(root)
    for n in nodes:
        uid = str(id(n))
        # for any value in the graph, create a rectangular ('record') node for it
        label = getattr(n, 'label', '')
        grad = getattr(n, 'grad', None)
        text = f"{label} | data {n.data:.4f}" + (f" | grad {grad:.4f}" if grad is not None else "")
        dot.node(name=uid, label="{ " + text + " }", shape='record')
        if n.op:
            # if this value is a result of some operation, create an op node for it
            dot.node(name=uid + n.op, label=n.op)
            # and connect this node to it
            dot.edge(uid + n.op, uid)

    for n1, n2 in edges:
        # connect n1 to the op node of n2
        dot.edge(str(id(n1)), str(id(n2)) + (n2.op or ''))

    return dot


g.grad = -0.7

g.backward()

# draw_dot(g)
```

```python
class Neuron():
    def __init__(self, n_inputs):
        self.weights = [Value(random.uniform(-1,1), label=f"w{i}") for i in range(n_inputs)]
        self.b = Value(0.0,label='b')

    def __call__(self, x):  # X is a vector of size n_inputs
        activation_value = sum((wi*xi for wi,xi in zip(self.weights, x)), self.b)
        return activation_value

    def parameters(self):
        return self.weights + [self.b]

class Layer():
    def __init__(self, n_neurons, n_inputs):
        self.neurons = [Neuron(n_inputs) for i in range(n_neurons)]

    def __call__(self, x):
        out = [n(x) for n in self.neurons]
        return out
    
    def parameters(self):
        return [p for n in self.neurons for p in n.parameters()]
    
    
n1 = Neuron(3)
x = [
    Value(1.3,label='x0'),
    Value(2.5,label='x1'),
    Value(1.3,label='x2')
]
l1 = Layer(4, 3)
l2 = Layer(4, 3)
out = n1(x)
draw_dot(out)
```

```python
class MLP():
    def __init__(self, n_inputs, n_outs):
        sizes = [n_inputs] + n_outs
        self.layers = [Layer(sizes[i+1], sizes[i]) for i in range(len(n_outs))]
        # Karpathy's trick: activation on every layer except the last one
        self.nonlin = [i != len(n_outs) - 1 for i in range(len(n_outs))]

    def __call__(self, x):
        for li, (layer, nonlin) in enumerate(zip(self.layers, self.nonlin)):
            x = layer(x)
            if nonlin:
                x = [v.swish() for v in x]
            for j, v in enumerate(x):
                v.label = f"L{li}.n{j}"
        return x[0] if len(x) == 1 else x


def draw_mlp(mlp, n_inputs):
    # neuron-level view: one node per neuron, edges = which outputs feed which neurons
    dot = Digraph(format='svg', graph_attr={'rankdir': 'LR', 'splines': 'line', 'nodesep': '0.3', 'ordering': 'in'})
    prev = [f"x{i}" for i in range(n_inputs)]
    for p in prev:
        dot.node(p, p, shape='circle')
    for li, (layer, nonlin) in enumerate(zip(mlp.layers, mlp.nonlin)):
        cur = []
        for j in range(len(layer.neurons)):
            name = f"L{li}.n{j}"
            act = 'swish' if nonlin else 'linear'
            color = 'lightblue' if nonlin else 'lightsalmon'
            dot.node(name, f"{name}\n{act}", shape='circle', style='filled', fillcolor=color)
            for p in prev:
                dot.edge(p, name)
            cur.append(name)
        prev = cur
    return dot


mlp = MLP(3, [4, 4, 1])
y = mlp(x)
draw_mlp(mlp, 3)
```

```python
# full scalar computation graph of the same network (every weight, *, + and swish)
y.grad = 1.0
y.backward()
draw_dot(y)
```

```python
class SwiGLULayer():
    # h = swish(W_gate·x) ⊙ (W_up·x): two linear layers on the same input, multiplied elementwise
    def __init__(self, n_neurons, n_inputs):
        self.gate = Layer(n_neurons, n_inputs)
        self.up = Layer(n_neurons, n_inputs)

    def __call__(self, x):
        g = [v.swish() for v in self.gate(x)]
        u = self.up(x)
        return [gi * ui for gi, ui in zip(g, u)]

    def parameters(self):
        return self.gate.parameters() + self.up.parameters()


class SwiGLUMLP():
    def __init__(self, n_inputs, n_outs):
        sizes = [n_inputs] + n_outs
        # same trick as before: hidden layers are gated, the last layer stays a plain linear projection
        self.layers = [SwiGLULayer(sizes[i+1], sizes[i]) if i != len(n_outs) - 1 else Layer(sizes[i+1], sizes[i])
                       for i in range(len(n_outs))]

    def __call__(self, x):
        for li, layer in enumerate(self.layers):
            x = layer(x)
            for j, v in enumerate(x):
                v.label = f"L{li}.h{j}"
        return x[0] if len(x) == 1 else x

    def parameters(self):
        return [p for layer in self.layers for p in layer.parameters()]


def draw_swiglu_mlp(mlp, n_inputs):
    # neuron-level view: each gated layer shows its gate neurons, up neurons and the ⊙ that joins them
    dot = Digraph(format='svg', graph_attr={'rankdir': 'LR', 'splines': 'line', 'nodesep': '0.2', 'ordering': 'in'})
    prev = [f"x{i}" for i in range(n_inputs)]
    for p in prev:
        dot.node(p, p, shape='circle')
    for li, layer in enumerate(mlp.layers):
        cur = []
        if isinstance(layer, SwiGLULayer):
            for j in range(len(layer.gate.neurons)):
                gate, up, h = f"L{li}.g{j}", f"L{li}.u{j}", f"L{li}.h{j}"
                dot.node(gate, f"{gate}\nswish", shape='circle', style='filled', fillcolor='lightblue')
                dot.node(up, f"{up}\nlinear", shape='circle', style='filled', fillcolor='lightgrey')
                dot.node(h, "⊙", shape='circle', style='filled', fillcolor='khaki', xlabel=h)
                for p in prev:
                    dot.edge(p, gate)
                    dot.edge(p, up)
                dot.edge(gate, h)
                dot.edge(up, h)
                cur.append(h)
        else:
            for j in range(len(layer.neurons)):
                name = f"L{li}.n{j}"
                dot.node(name, f"{name}\nlinear", shape='circle', style='filled', fillcolor='lightsalmon')
                for p in prev:
                    dot.edge(p, name)
                cur.append(name)
        prev = cur
    return dot


smlp = SwiGLUMLP(3, [4, 4, 1])
y2 = smlp(x)
draw_swiglu_mlp(smlp, 3)
```

```python
# backprop through the gated network, then sanity-check one gate weight against a numerical gradient
y2.grad = 1.0
y2.backward()

X = [
    [
        Value(1.3,label='x0_0'),
        Value(2.5,label='x0_1'),
        Value(1.3,label='x0_2')
    ],
    [
        Value(1,label='x1_0'),
        Value(3,label='x1_1'),
        Value(4,label='x1_2')
    ],
    [
        Value(2.3,label='x2_0'),
        Value(3.5,label='x2_1'),
        Value(1.3,label='x2_2')
    ],
    [
        Value(2.3,label='x3_0'),
        Value(3.5,label='x3_1'),
        Value(1.3,label='x3_2')
    ]
]

network = SwiGLUMLP(3, [4, 4, 1])

Y_targets = [
    Value(2.0, label='Y1'),
    Value(3.0, label='Y2'),
    Value(4.0, label='Y3'),
    Value(5.0, label='Y4')
]

# loss.grad = 1
# loss.backward()


def train_loop(LR, step):
    for i in range(step):
        # Forward:
        Y = [network(x) for x in X]
        loss = sum([(y - target) ** 2 for y, target in zip(Y, Y_targets)])

        # Backward
        loss.grad = 1
        loss.backward()
        
        for p in network.parameters():
            p.data += -LR * p.grad

        print(f'step {i}: loss: {loss.data}. LR = {LR:.4f}')

# draw_dot(loss)
train_loop(0.01, 20)
```
