```python
from collections import deque
import numpy as np


class Value():
    def __init__(self, data, children=[], op=None):
        self.data = data
        self.children = children
        self.op = op
        self.label = ''
        self.grad = 0.0
        # self._backward = None

    def __add__(self, other):
        out = Value(self.data + other.data, [self, other], '+')

        def _backward():
            self.grad += out.grad
            other.grad += out.grad
        
        out._backward = _backward
        return out

    def __mul__(self, other):
        out = Value(self.data * other.data, [self, other], '*')

        def _backward():
            self.grad += other.data * out.grad
            other.grad += self.data * out.grad

        out._backward = _backward
        return out

    def sigmoid(self):
        out = Value(1 / (1 + np.exp(-self.data)), [self], 'sigmoid')

        def _backward():
            self.grad += out.data * (1 - out.data) * out.grad

        out._backward = _backward
        return out

    def __repr__(self):
        return f"Value(data={self.data},children={len(self.children)},op={self.op})"

    def _backward(self):
        pass # leaf node
        # raise Exception("Not implemented")       

    def backward(self):
        # count how many edges lead into each node from above (its consumers),
        # i.e. how many gradient contributions it has to wait for
        pending = {}
        seen = {self}
        queue = deque([self])
        while queue:
            node = queue.popleft()
            for child in node.children:
                pending[child] = pending.get(child, 0) + 1
                if child not in seen:
                    seen.add(child)
                    queue.append(child)

        # zero every grad below the root so re-running backward doesn't accumulate
        for node in seen:
            if node is not self:
                node.grad = 0.0

        # BFS from the root, but only enqueue a node once all its consumers have
        # pushed their gradient into it (Kahn's algorithm), so its grad is final
        queue = deque([self])
        while queue:
            node = queue.popleft()
            node._backward()
            for child in node.children:
                pending[child] -= 1
                if pending[child] == 0:
                    queue.append(child)



a = Value(3.0); a.label='a'
b = Value(2.0); b.label='b'
bs = b.sigmoid(); bs.label = 'bs'
c = a * bs; c.label='c'
f = Value(3.0); f.label='f'
e = c * f; e.label='e'
g  = e + f; g.label='g'
h = g.sigmoid(); h.label = 'h'

g
```

```python
from graphviz import Digraph

def trace(root):
    # builds a set of all nodes and edges in a graph
    nodes, edges = set(), set()
    def build(v):
        if v not in nodes:
            nodes.add(v)
            for child in v.children:
                edges.add((child, v))
                build(child)
    build(root)
    return nodes, edges

def draw_dot(root):
    dot = Digraph(format='svg', graph_attr={'rankdir': 'LR'}) # LR = left to right

    nodes, edges = trace(root)
    for n in nodes:
        uid = str(id(n))
        # for any value in the graph, create a rectangular ('record') node for it
        label = getattr(n, 'label', '')
        grad = getattr(n, 'grad', None)
        text = f"{label} | data {n.data:.4f}" + (f" | grad {grad:.4f}" if grad is not None else "")
        dot.node(name=uid, label="{ " + text + " }", shape='record')
        if n.op:
            # if this value is a result of some operation, create an op node for it
            dot.node(name=uid + n.op, label=n.op)
            # and connect this node to it
            dot.edge(uid + n.op, uid)

    for n1, n2 in edges:
        # connect n1 to the op node of n2
        dot.edge(str(id(n1)), str(id(n2)) + (n2.op or ''))

    return dot


g.grad = -0.7

g.backward()

draw_dot(g)
```
