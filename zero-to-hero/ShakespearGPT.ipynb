```python
# Shakespear data from https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt

with open('input.txt', 'r', encoding='utf-8') as f:
    text = f.read()

# print(len(text))
charset = sorted(list(set(text)))
print(len(charset))
encode_map = { ch:i for i,ch in enumerate(charset) }
decode_map = { i:ch for i,ch in enumerate(charset) }

def encode(text):
    return [encode_map[char] for char in text]

def decode(index_arr):
    return ''.join([decode_map[i] for i in index_arr])

# print(encode("hii there"))
# print(decode(encode("hii there")))

import torch # we use PyTorch: https://pytorch.org
data = torch.tensor(encode(text), dtype=torch.long)
print(data.shape, data.dtype)
print(data[:15])

n = len(data) * 0.9
train_data = data[:n]
val_data = data[n:]

```
